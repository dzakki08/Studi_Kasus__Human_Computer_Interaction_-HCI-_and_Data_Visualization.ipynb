## 4. Konsep HCI dan Visualisasi Data

**Human-Computer Interaction (HCI)** mempelajari perancangan, implementasi, dan evaluasi sistem interaktif yang digunakan manusia. Tiga komponen utama HCI adalah manusia, komputer/sistem, dan cara keduanya berinteraksi.

**Visualisasi data** menyajikan data melalui grafik atau representasi visual agar pola dan perbandingan lebih mudah dipahami.

Prinsip HCI yang digunakan:
- **Visibility:** status dan informasi utama mudah ditemukan.
- **Feedback:** sistem memberi respons setelah pengguna melakukan tindakan.
- **Consistency:** pola navigasi, label, dan kontrol digunakan secara konsisten.
- **Affordance:** tampilan kontrol memberi petunjuk cara menggunakannya.
- **Constraint:** membatasi tindakan yang tidak sesuai atau tidak valid.

## 9. Prinsip HCI yang Diterapkan

- **Visibility:** kartu ringkasan menampilkan metrik utama di bagian atas.
- **Feedback:** periode yang dipilih terlihat jelas dan perubahan filter diperbarui.
- **Consistency:** format tanggal, angka, label, dan navigasi seragam.
- **Affordance:** filter terlihat seperti kontrol yang dapat digunakan.
- **Constraint:** periode laporan hanya dapat dipilih dari rentang yang tersedia.
- **Error prevention:** pesan validasi ditampilkan jika periode tidak valid.
- **Recognition rather than recall:** label kategori ditampilkan langsung agar pengguna tidak perlu mengingat kode.

## 10. Menyiapkan Data Simulasi

Data simulasi memuat catatan bulanan untuk beberapa kategori buku. Variabel meliputi jumlah peminjaman, pengunjung, dan pengembalian terlambat. Nilai dibuat untuk demonstrasi visualisasi dan bukan hasil pengukuran perpustakaan nyata.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(2026)
months = pd.date_range("2026-01-01", periods=8, freq="MS")
categories = ["Teknologi", "Sains", "Sastra", "Bisnis", "Sejarah"]

rows = []
for i, month in enumerate(months):
    monthly_visitors = int(rng.integers(700, 1500))
    for category in categories:
        base = {"Teknologi": 180, "Sains": 130, "Sastra": 160, "Bisnis": 110, "Sejarah": 90}[category]
        seasonal = 12 * i
        borrowed = int(max(15, base + seasonal + rng.normal(0, 22)))
        late_returns = int(max(0, borrowed * rng.uniform(0.04, 0.16)))
        rows.append({
            "bulan": month,
            "kategori_buku": category,
            "jumlah_peminjaman": borrowed,
            "jumlah_pengunjung": monthly_visitors,
            "pengembalian_terlambat": late_returns
        })

df = pd.DataFrame(rows)
print("Lima baris pertama:")
display(df.head())
print("Ukuran dataset:", df.shape)
```

## 11. Memeriksa Data

```python
print("Ukuran data:", df.shape)
print("\\nTipe data:")
print(df.dtypes)
print("\\nNilai kosong:")
display(df.isna().sum().to_frame("jumlah_missing"))
print("Jumlah baris duplikat:", df.duplicated().sum())
print("\\nRingkasan statistik:")
display(df[["jumlah_peminjaman", "jumlah_pengunjung", "pengembalian_terlambat"]].describe().round(2))
```

## 12. Visualisasi 1 — Perbandingan Kategori Buku

Grafik batang cocok untuk membandingkan jumlah peminjaman antar kategori. Semua kategori menggunakan skala yang sama agar perbandingan tidak menyesatkan.

```python
avg_category = (
    df.groupby("kategori_buku", as_index=False)["jumlah_peminjaman"]
      .mean()
      .sort_values("jumlah_peminjaman", ascending=False)
)

plt.figure(figsize=(8, 4.5))
plt.bar(avg_category["kategori_buku"], avg_category["jumlah_peminjaman"])
plt.title("Rata-rata Peminjaman per Kategori Buku")
plt.xlabel("Kategori buku")
plt.ylabel("Rata-rata peminjaman per bulan")
plt.tight_layout()
plt.show()
display(avg_category.round(2))
```

**Interpretasi:** kategori dengan rata-rata peminjaman lebih tinggi dapat menjadi bahan pertimbangan untuk mengevaluasi kebutuhan koleksi. Namun, jumlah peminjaman tidak otomatis menunjukkan kualitas koleksi atau kebutuhan semua pengunjung.

## 13. Visualisasi 2 — Tren Peminjaman dari Waktu ke Waktu

Grafik garis membantu melihat perubahan jumlah peminjaman sepanjang bulan. Tren perlu dibaca bersama konteks seperti kalender akademik, masa ujian, dan perubahan jam layanan.

```python
monthly_total = df.groupby("bulan", as_index=False)["jumlah_peminjaman"].sum()

plt.figure(figsize=(9, 4.5))
plt.plot(monthly_total["bulan"], monthly_total["jumlah_peminjaman"], marker="o")
plt.title("Tren Total Peminjaman per Bulan")
plt.xlabel("Bulan")
plt.ylabel("Jumlah peminjaman")
plt.xticks(monthly_total["bulan"], monthly_total["bulan"].dt.strftime("%b %Y"), rotation=30)
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
display(monthly_total)
```

**Interpretasi:** grafik membantu mengenali kenaikan atau penurunan peminjaman. Grafik tidak menjelaskan penyebab perubahan; petugas perlu memeriksa kalender akademik dan informasi layanan sebelum menarik kesimpulan.

## 14. Visualisasi 3 — Pengunjung dan Pengembalian Terlambat

Jumlah pengunjung diringkas per bulan karena nilainya sama pada setiap kategori di bulan yang sama. Jumlah pengembalian terlambat dijumlahkan untuk melihat tren keterlambatan secara keseluruhan.

```python
monthly_summary = df.groupby("bulan", as_index=False).agg(
    jumlah_pengunjung=("jumlah_pengunjung", "max"),
    pengembalian_terlambat=("pengembalian_terlambat", "sum")
)

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.bar(monthly_summary["bulan"].dt.strftime("%b %Y"), monthly_summary["jumlah_pengunjung"])
ax.set_title("Jumlah Pengunjung Perpustakaan per Bulan (Simulasi)")
ax.set_xlabel("Bulan")
ax.set_ylabel("Jumlah pengunjung")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(monthly_summary["bulan"].dt.strftime("%b %Y"),
        monthly_summary["pengembalian_terlambat"], marker="o")
ax.set_title("Pengembalian Terlambat per Bulan (Simulasi)")
ax.set_xlabel("Bulan")
ax.set_ylabel("Jumlah pengembalian terlambat")
plt.xticks(rotation=30)
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
display(monthly_summary)
```

## 17. Prototipe dan Evaluasi Usability

Desain dapat dikembangkan secara bertahap:
- **Low-fidelity:** sketsa posisi kartu metrik, grafik, dan filter.
- **High-fidelity:** rancangan digital dengan label, warna, tipografi, dan tata letak yang mendekati tampilan akhir.
- **Uji usability:** minta calon pengguna menyelesaikan tugas seperti menemukan kategori paling banyak dipinjam atau melihat bulan dengan keterlambatan tertinggi.
- **Pengukuran:** catat keberhasilan tugas, waktu penyelesaian, kesalahan, dan penilaian kemudahan pengguna.
- **Iterasi:** perbaiki desain berdasarkan masalah yang ditemukan.

Uji coba harus dilakukan dengan partisipan yang sesuai dan tidak mengumpulkan data pribadi yang tidak diperlukan.

## 18. Heuristik Nielsen untuk Evaluasi

Sepuluh heuristik Nielsen dapat digunakan sebagai daftar periksa:
1. Visibility of system status.
2. Match between system and the real world.
3. User control and freedom.
4. Consistency and standards.
5. Error prevention.
6. Recognition rather than recall.
7. Flexibility and efficiency of use.
8. Aesthetic and minimalist design.
9. Help users recognize, diagnose, and recover from errors.
10. Help and documentation.

Contoh penerapan: saat filter periode berubah, dashboard menampilkan periode aktif; jika tidak ada data, sistem menampilkan pesan yang jelas; label kategori memakai istilah yang familiar bagi petugas.

## 19. Aksesibilitas, Dark Patterns, dan Etika

- Gunakan ukuran teks yang mudah dibaca dan kontras warna memadai.
- Sediakan label yang jelas untuk tombol dan kontrol.
- Jangan menggunakan warna sebagai satu-satunya pembeda.
- Pastikan navigasi dapat dipahami dan kontrol dapat digunakan dengan berbagai perangkat input.
- Hindari dark patterns, misalnya menyembunyikan filter penting atau membuat pengguna sulit membatalkan pilihan.
- Lindungi informasi identitas peminjam; tampilkan data agregat bila detail individu tidak diperlukan.
